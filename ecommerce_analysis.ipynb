```python
events_mysql = events.copy()

# Columns that should be integers but contain missing values
events_mysql['product_id'] = events_mysql['product_id'].astype('Int64')
events_mysql['qty'] = events_mysql['qty'].astype('Int64')
events_mysql['cart_size'] = events_mysql['cart_size'].astype('Int64')
```

```python
events_mysql = events_mysql.where(pd.notna(events_mysql), None)
```
